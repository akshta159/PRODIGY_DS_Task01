```python
import pandas as pd
import matplotlib.pyplot as plt

url ="https://raw.githubusercontent.com/Prodigy-InfoTech/data-science-datasets/main/Task%201/API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv"
df = pd.read_csv(url, skiprows=4)
print(df.head())
print(df.columns)
```

```python
plt.hist(df['2020'].dropna(), bins=20, color='skyblue', edgecolor='black')
plt.title("Population Distribution (2020)")
plt.xlabel("Population")
plt.ylabel("Number of Countries")
plt.show()
```

```python
top10 = df[['Country Name','2020']].sort_values(by='2020', ascending=False).head(10)

plt.bar(top10['Country Name'], top10['2020'], color='green')
plt.xticks(rotation=45, ha='right')
plt.title("Top 10 Countries by Population (2020)")
plt.xlabel("Country")
plt.ylabel("Population")
plt.show()
```
